```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
from IPython.display import display

# Find the uploaded files
def find_file(keyword):
    matches = [
        name for name in uploaded
        if keyword.lower() in Path(name).stem.lower()
    ]

    if len(matches) != 1:
        raise ValueError(
            f"Expected one file matching '{keyword}', found: {matches}"
        )

    return matches[0]

p1_gen_file = find_file("Plant_1_Generation_Data")
p1_weather_file = find_file("Plant_1_Weather_Sensor_Data")
p2_gen_file = find_file("Plant_2_Generation_Data")
p2_weather_file = find_file("Plant_2_Weather_Sensor_Data")

# Load the CSV files
plant1_generation = pd.read_csv(p1_gen_file)
plant1_weather = pd.read_csv(p1_weather_file)
plant2_generation = pd.read_csv(p2_gen_file)
plant2_weather = pd.read_csv(p2_weather_file)

# Display basic information
datasets = {
    "Plant 1 Generation": plant1_generation,
    "Plant 1 Weather": plant1_weather,
    "Plant 2 Generation": plant2_generation,
    "Plant 2 Weather": plant2_weather
}

for name, df in datasets.items():
    print(f"\n{'=' * 50}")
    print(name)
    print("Shape:", df.shape)
    print("Columns:", df.columns.tolist())
    display(df.head())
```

```python
import pandas as pd
import numpy as np

# Step 1: Standardize column names
dataframes = [
    plant1_generation,
    plant2_generation,
    plant1_weather,
    plant2_weather
]

for df in dataframes:
    df.columns = df.columns.str.strip().str.upper()

# Step 2: Remove previously added plant identifiers
for df in dataframes:
    if "PLANT_ID" in df.columns:
        df.drop(columns=["PLANT_ID"], inplace=True)

# Step 3: Assign plant identifiers
plant1_generation["PLANT_ID"] = 1
plant2_generation["PLANT_ID"] = 2

plant1_weather["PLANT_ID"] = 1
plant2_weather["PLANT_ID"] = 2

# Step 4: Combine data from both plants
generation = pd.concat(
    [plant1_generation, plant2_generation],
    ignore_index=True
)

weather = pd.concat(
    [plant1_weather, plant2_weather],
    ignore_index=True
)

# Step 5: Convert timestamps
generation["DATE_TIME"] = pd.to_datetime(
    generation["DATE_TIME"],
    errors="coerce"
)

weather["DATE_TIME"] = pd.to_datetime(
    weather["DATE_TIME"],
    errors="coerce"
)

# Step 6: Remove duplicate records
generation = generation.drop_duplicates().copy()
weather = weather.drop_duplicates().copy()

# Step 7: Remove records with invalid timestamps
generation = generation.dropna(
    subset=["DATE_TIME"]
).copy()

weather = weather.dropna(
    subset=["DATE_TIME"]
).copy()

# Step 8: Sort data by plant and timestamp
generation = generation.sort_values(
    by=["PLANT_ID", "DATE_TIME"]
).reset_index(drop=True)

weather = weather.sort_values(
    by=["PLANT_ID", "DATE_TIME"]
).reset_index(drop=True)

# Step 9: Display results
print("=" * 50)
print("DATA CLEANING COMPLETED")
print("=" * 50)

print("\nGeneration shape:", generation.shape)
print("Weather shape:", weather.shape)

print("\nGeneration columns:")
print(generation.columns.tolist())

print("\nWeather columns:")
print(weather.columns.tolist())

print("\nGeneration records per plant:")
print(generation["PLANT_ID"].value_counts().sort_index())

print("\nWeather records per plant:")
print(weather["PLANT_ID"].value_counts().sort_index())

print("\nGeneration preview:")
display(generation.head())

print("\nWeather preview:")
display(weather.head())
```

```python
# 1. Check missing values
print("Missing values in Generation Data:")
display(generation.isnull().sum().to_frame("Missing Values"))

print("\nMissing values in Weather Data:")
display(weather.isnull().sum().to_frame("Missing Values"))

# 2. Check duplicate records
print("\nDuplicate rows in Generation:", generation.duplicated().sum())
print("Duplicate rows in Weather:", weather.duplicated().sum())

# 3. Check data types
print("\nGeneration Data Types:")
print(generation.dtypes)

print("\nWeather Data Types:")
print(weather.dtypes)

# 4. Check date ranges
print("\nGeneration Date Range:")
print(generation["DATE_TIME"].min(), "to", generation["DATE_TIME"].max())

print("\nWeather Date Range:")
print(weather["DATE_TIME"].min(), "to", weather["DATE_TIME"].max())
```

```python
print("Generation Data Statistics:")
display(generation.describe().round(2))

print("\nWeather Data Statistics:")
display(weather.describe().round(2))

# Compare average and maximum power by plant
plant_analysis = (
    generation.groupby("PLANT_ID")
    .agg(
        Average_AC_Power=("AC_POWER", "mean"),
        Maximum_AC_Power=("AC_POWER", "max"),
        Average_DC_Power=("DC_POWER", "mean"),
        Number_of_Inverters=("SOURCE_KEY", "nunique")
    )
    .round(2)
    .reset_index()
)

display(plant_analysis)
```

```python
plant_analysis.plot(
    x="PLANT_ID",
    y=["Average_AC_Power", "Maximum_AC_Power"],
    kind="bar",
    figsize=(9, 5)
)

plt.title("Solar Power Generation by Plant")
plt.xlabel("Plant ID")
plt.ylabel("Power (kW)")
plt.xticks(rotation=0)
plt.legend(["Average AC Power", "Maximum AC Power"])
plt.tight_layout()
plt.show()
```

```python
generation["HOUR"] = generation["DATE_TIME"].dt.hour

hourly_analysis = (
    generation.groupby(["PLANT_ID", "HOUR"])["AC_POWER"]
    .mean()
    .reset_index()
)

plt.figure(figsize=(12, 5))

for plant_id, data in hourly_analysis.groupby("PLANT_ID"):
    plt.plot(
        data["HOUR"],
        data["AC_POWER"],
        marker="o",
        label=f"Plant {plant_id}"
    )

plt.title("Average Solar Power by Hour")
plt.xlabel("Hour of Day")
plt.ylabel("Average AC Power (kW)")
plt.xticks(range(24))
plt.grid(alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
generation["DATE"] = generation["DATE_TIME"].dt.date

daily_inverter_energy = (
    generation.groupby(["PLANT_ID", "DATE", "SOURCE_KEY"])
    ["DAILY_YIELD"]
    .max()
    .reset_index()
)

daily_production = (
    daily_inverter_energy.groupby(["PLANT_ID", "DATE"])
    ["DAILY_YIELD"]
    .sum()
    .reset_index(name="Estimated_Daily_Energy")
)

daily_production["DATE"] = pd.to_datetime(daily_production["DATE"])

plt.figure(figsize=(13, 5))

for plant_id, data in daily_production.groupby("PLANT_ID"):
    plt.plot(
        data["DATE"],
        data["Estimated_Daily_Energy"],
        marker="o",
        label=f"Plant {plant_id}"
    )

plt.title("Daily Solar Energy Production")
plt.xlabel("Date")
plt.ylabel("Energy (dataset yield units)")
plt.xticks(rotation=45)
plt.grid(alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()

display(daily_production.head(10))
```

```python
# Aggregate inverter readings by plant and timestamp
generation_by_time = (
    generation.groupby(
        ["PLANT_ID", "DATE_TIME"],
        as_index=False
    )
    .agg(
        AC_POWER=("AC_POWER", "sum"),
        DC_POWER=("DC_POWER", "sum")
    )
)

print("Aggregated generation shape:", generation_by_time.shape)
display(generation_by_time.head())
```

```python
# Select weather columns
weather_columns = [
    "PLANT_ID",
    "DATE_TIME",
    "AMBIENT_TEMPERATURE",
    "MODULE_TEMPERATURE",
    "IRRADIATION"
]

# Keep only columns that exist
weather_columns = [
    col for col in weather_columns
    if col in weather.columns
]

weather_clean = weather[weather_columns].copy()

# Merge generation and weather measurements
merged = generation_by_time.merge(
    weather_clean,
    on=["PLANT_ID", "DATE_TIME"],
    how="left",
    validate="one_to_one"
)

print("Merged dataset shape:", merged.shape)
print("\nMissing values after merge:")
display(merged.isnull().sum().to_frame("Missing Values"))

display(merged.head())
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(9, 5))

for plant_id, data in merged.groupby("PLANT_ID"):
    plt.scatter(
        data["IRRADIATION"],
        data["AC_POWER"],
        alpha=0.4,
        label=f"Plant {plant_id}"
    )

plt.title("Solar Irradiation vs AC Power")
plt.xlabel("Irradiation (source units)")
plt.ylabel("AC Power (kW)")
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(9, 5))

plt.scatter(
    merged["MODULE_TEMPERATURE"],
    merged["AC_POWER"],
    alpha=0.4
)

plt.title("Module Temperature vs AC Power")
plt.xlabel("Module Temperature (°C)")
plt.ylabel("AC Power (kW)")
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
correlation_columns = [
    "AC_POWER",
    "DC_POWER",
    "AMBIENT_TEMPERATURE",
    "MODULE_TEMPERATURE",
    "IRRADIATION"
]

correlation_columns = [
    col for col in correlation_columns
    if col in merged.columns
]

correlation_matrix = merged[correlation_columns].corr()

display(correlation_matrix.round(3))
```

```python
# Use high-irradiation observations only
irradiation_threshold = merged["IRRADIATION"].quantile(0.75)

daylight_data = merged[
    merged["IRRADIATION"] > irradiation_threshold
].copy()

# Calculate the lower 10th-percentile power threshold per plant
daylight_data["LOW_OUTPUT_THRESHOLD"] = (
    daylight_data.groupby("PLANT_ID")["AC_POWER"]
    .transform(lambda x: x.quantile(0.10))
)

low_output = daylight_data[
    daylight_data["AC_POWER"]
    < daylight_data["LOW_OUTPUT_THRESHOLD"]
].copy()

print("Potential low-output observations:", len(low_output))

display(
    low_output[
        [
            "PLANT_ID",
            "DATE_TIME",
            "AC_POWER",
            "IRRADIATION",
            "LOW_OUTPUT_THRESHOLD"
        ]
    ].head(20)
)
```
